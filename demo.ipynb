## 2. Мәтін статистикасы (кілт керек емес)
`text` ішіне кез келген мәтін қой.

```python
from textstats import stats

text = """Жасанды интеллект — компьютерлерге адамның ойлау қабілетіне ұқсас міндеттерді орындауға мүмкіндік беретін технология. Қазір ол аударма, іздеу жүйелері және дауыстық көмекшілерде қолданылады."""
s = stats(text)
print("Сөз:", s["words"], "| Сөйлем:", s["sentences"], "| Оқу уақыты:", s["reading_minutes"], "мин")
print("Жиі сөздер:", s["top_words"])
```
